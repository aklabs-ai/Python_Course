## 1. The `if` Statement

The `if` statement runs a block of code **only if** its condition evaluates to `True`.
Python uses **indentation** (not braces `{}`) to define the block of code that belongs to the `if`.

```python
if condition:
    # code runs only if condition is True
```

````python
age = 20

if age >= 18:
    print("You are eligible to vote.")

print("This line always runs, regardless of the condition.")
````

## 2. The `if-else` Statement

`else` provides an alternate block of code that runs when the `if` condition is `False`.

```python
if condition:
    # runs if True
else:
    # runs if False
```

````python
age = 15

if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are NOT eligible to vote yet.")
````

## 3. The `if-elif-else` Ladder

`elif` (short for "else if") lets you check **multiple conditions** in sequence. Python checks each
condition top to bottom and runs the block for the **first** one that is `True`; if none are `True`,
the final `else` block runs.

```python
if condition1:
    ...
elif condition2:
    ...
elif condition3:
    ...
else:
    ...
```

````python
marks = 72

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 40:
    grade = "D"
else:
    grade = "Fail"

print(f"Marks: {marks} -> Grade: {grade}")   


requested_toppings = ['mushrooms', 'extra cheese']

if 'mushrooms' in requested_toppings:
    print("Adding mushrooms.")
elif 'pepperoni' in requested_toppings:
    print("Adding pepperoni.")
elif 'extra cheese' in requested_toppings:
    print("Adding extra cheese.")
print("\nFinished making your pizza!")
````

## 4. Nested `if` Statements

An `if` statement can be placed **inside** another `if` (or `else`) block. This is useful when a
decision depends on more than one level of conditions.

````python
num = 15

if num > 0:
    print("Number is positive")
    if num % 2 == 0:
        print("It is also even")
    else:
        print("It is also odd")
else:
    print("Number is zero or negative")
````

## 5. Short-Hand `if` and the Ternary Operator

For simple conditions, Python offers compact one-line syntax.

**Short-hand if** (only when there's no `else`):
```python
if condition: statement
```

**Ternary (conditional) expression** — assigns a value based on a condition in a single line:
```python
value = true_result if condition else false_result
```

````python
# Short-hand if
age = 20
if age >= 18: print("Adult")

# Ternary operator
age = 15
status = "Adult" if age >= 18 else "Minor"
print(status)

# Ternary is an expression -> can be used directly inside print()
num = 7
print("Even" if num % 2 == 0 else "Odd")
````

## 6. Logical Operators in Conditions

Combine multiple conditions using `and`, `or`, and `not`.

| Operator | True when |
|---|---|
| `and` | **both** conditions are True |
| `or` | **at least one** condition is True |
| `not` | reverses the condition |

````python
age = 25
has_ticket = True

if age >= 18 and has_ticket:
    print("Entry allowed")
else:
    print("Entry denied")

temperature = 40
if temperature > 35 or temperature < 5:
    print("Extreme weather warning")
else:
    print("Weather is normal")

is_raining = False
if not is_raining:
    print("No need for an umbrella")
````

## 7. Truthy & Falsy Values in Conditions

In Python, `if` doesn't require a strict boolean — it evaluates the **"truthiness"** of any value.

**Falsy values** (treated as `False`): `0`, `0.0`, `""` (empty string), `[]`, `{}`, `()`, `None`, `False`

**Everything else is Truthy** (treated as `True`), including non-empty strings, non-zero numbers,
and non-empty collections.

````python
values = [0, 1, "", "hello", [], [1, 2], None, -5]

for v in values:
    if v:
        print(f"{v!r:>10} -> Truthy")
    else:
        print(f"{v!r:>10} -> Falsy")
````

## 8. Common Mistakes to Avoid

1. **Using `=` instead of `==`** — `=` is assignment, `==` is comparison.
   ```python
   if x = 5:   # SyntaxError! use if x == 5:
   ```
2. **Forgetting the colon `:`** at the end of `if`/`elif`/`else`.
3. **Inconsistent indentation** — Python is strict about indentation; mixing tabs and spaces causes errors.
4. **Comparing floats with `==` directly** — due to floating-point precision, prefer a tolerance check,
   e.g. `abs(a - b) < 1e-9`, instead of `a == b`.
5. **Using `if x == True:`** — just write `if x:` since `x` is already a boolean.

````python
# Demonstration of float comparison issue
a = 0.1 + 0.2
print("0.1 + 0.2 == 0.3 :", a == 0.3)          # False! due to floating point precision
print("Actual value      :", a)

# Correct way -> use a small tolerance
print("Using tolerance    :", abs(a - 0.3) < 1e-9)
````
